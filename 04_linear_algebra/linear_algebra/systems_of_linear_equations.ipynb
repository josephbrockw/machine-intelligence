# Systems of Linear Equations
An equation is linear if all terms are of degree 0 or 1. An equation can any number of variables.

A system is any number of linear equations. A solution of a system of linear equations is a set of points that make every equation true.


````python
# x1 + 2x2 = 3
# x1 - x2 = 0
````

**KEY IDEA:** solutions of linear systems correspond to how lines can meet.

How can lines meet?
1. Lines intersect (1 solution)
2. Lines are parallel (No solutions)
3. Lines are the same (Infinite solutions)

**DEF:** A linear system is _consistent_ if either 1 or infinite solutions exist, otherwise it is _inconsistent_

## Example
```
4x1 + 5x2 + 3x3 + 3x4 = 1
 x1 +  x2 +  x3 +  x4 = 0
2x1 + 3x2 +  x3 +  x4 = 1
5x1 + 7x2 + 3x3 + 3x4 = 2
```
To solve without graphing, create a _coefficient matrix_:
```text
4 5 3 3
1 1 1 1
2 3 1 1
5 7 3 3
```

or an _augmented matrix_ by adding the result as an additional column:
```text
4 5 3 3 1
1 1 1 1 0
2 3 1 1 1
5 7 3 3 2
```
**Order matters!**

**DEF:** Size of a matrix is the number of rows x number of cols

Coefficient Matrix: 4x4

Augmented Matrix: 4x5

## Elementary Row Operations (ERO)
1. Replacement - Replace one row by the sum of itself & multiple of another row (ex: 3R1 + R2)
2. Interchange - Swap the position of two rows (ex: R1 <-> R2)
3. Scaling     - Multiply all entries in a row by a non-zero constant (ex: 2 * R1)

**DEF:** Two matrices are row-equivalent if there is a sequence of EROs that transforms one into the other (A ~ ... ~ B)

_All EROs are reversible_

_EROs do not change the solution set of a linear system_

Given a Linear System:
1. Is a system consistent? (at least one solution)
2. If consistent, is solution unique? (1 or infinite)

## Solving with Substitution
Thus, you can use the following general steps to solve a system of linear equations.

Write the augmented matrix for the system of equations.

Perform elementary row operations until the matrix is in triangular form (that is, all the entries below the main diagonal are 0’s).

Interpret the new row equivalent matrix to determine the solutions. In general, start with the lowest row to find a value for one of the variables, then substitute that value into the equation above it to determine the value of the next variable, and so on.
