# Chain-of-Thought Prompting
## Load Pretrained Model

```python
import torch
from mi.lm.model_loader import load_model_and_tokenizer

# device = get_device()
device = torch.device("cpu")

model, tokenizer = load_model_and_tokenizer(
    which_model="base",
    device=device,
    use_compile=True
)
```

```python
from mi.lm.prompts import render_prompt

raw_prompt = (
    "Half the value of $3x-9$ is $x+37$. "
    "What is the value of $x$?"
)

prompt = render_prompt(raw_prompt)
print(prompt)
```

```python
from mi.lm.generate import (
    generate_text_stream_concat_flex,
    generate_text_basic_stream_cache
)

response = generate_text_stream_concat_flex(
    model, tokenizer, prompt, device,
    max_new_tokens=2048, verbose=True,
    generate_func=generate_text_basic_stream_cache
)
```

## Chain-of-Thought Prompting

```python
prompt_cot = prompt + "\n\nExplain step by step."

response_cot = generate_text_stream_concat_flex(
    model, tokenizer, prompt_cot, device,
    max_new_tokens=2048, verbose=True
)
```

### Chain-of-Thought Reasoning

How COT Reasoning can help:

1. Walking through the steps gives the model more opportunities to correct itself.
2. Step-by-step reasoning matches the way many training examples are written. For instance, large math and logic datasets often contain detailed solutions, so asking for a chain of thought aligns the model with patterns it has already learned.

Overall, chain-of-thought prompting does not provide the model with new knowledge, but it changes how the model uses its existing knowledge. Often, this shift can lead to more reliable answers. This is especially true for math, code, logic, and other multistep problems.
