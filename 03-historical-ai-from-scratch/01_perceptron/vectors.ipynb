```python
a = [4, 3]
b = [2, 6]

# Adding two vectors
result = [a[i] + b[i] for i in range(len(a))]
print(result)
```

```python
# Subtracting two vectors
result = [a[i] - b[i] for i in range(len(a))]
print(result)
```

```python
# Scaling a vector
result = [5 * a[i] for i in range(len(a))]
print(result)
```

```python
import math
a_magnitude = math.sqrt(pow(a[0], 2) + pow(a[1], 2))
print(f"Original magnitude: {a_magnitude}")
scaled_magnitude = math.sqrt(pow(result[0], 2) + pow(result[1], 2))
print(f"Scaled magnitude: {scaled_magnitude}")
```

## Dot Product


```python
a = [4, 0]
b = [5, 5]
a_magnitude = math.sqrt(pow(a[0], 2) + pow(a[1], 2))
b_magnitude = math.sqrt(pow(b[0], 2) + pow(b[1], 2))

a_dot_b = a_magnitude * b_magnitude * math.cos(math.pi / 4)
print(f"a.b: {a_dot_b}")
```

If one of the vectors is a unit vector (magnitude of 1), the dot product is equal to the projection of the other vector onto the vector of unit length.

```python
def magnitude(vector):
    return math.sqrt(sum([pow(x, 2) for x in vector]))
```

```python
a = [1, 0]
b = [3, 3]

a_dot_b = magnitude(a) * magnitude(b) * math.cos(math.pi / 4)
print(f"a.b: {a_dot_b}")
```

Still true, even if the vector isn't on an axis.

```python
a = [1/math.sqrt(2), 1/math.sqrt(2)]
b = [1, 3]
```

```python
def dot_product(vector1, vector2):
    return sum(x * y for x, y in zip(vector1, vector2))
```

```python
dot_product(a, b)
```
