## A Model of a Biological Neuron
A 1943 paper by McCulloch and Pitts proposed a model of a neuron which they turned into a computational model, an artificial neuron. They showed that using these artificial neurons would allow you to implement basic boolean logical operations.


```python
def sum_inputs(inputs):
    return sum(inputs)


def artificial_neuron(inputs, threshold=0):
    summed_inputs = sum_inputs(inputs)
    if summed_inputs < threshold:
        return 0
    else:
        return 1
```

```python
print(artificial_neuron([1, 0, 0, 0, 1], 3))
print(artificial_neuron([0, 1, 0, 1, 1], 3))
```

The limitation is that there is no way for the system to learn what the right threshold is. That has to be set by the person.

## An Adjustable Artificial Neuron

Rosenblatt's big additions to the model of a neuron (which he referred to as a cell) was that inputs weren't limited to being binary and adding a weight to each input in a cell that would influence how amplifying or inhibitory that input would be to the cells further in the network.

```python
def weighted_sum(inputs, weights):
    summed_inputs = sum([w * x for x, w in zip(inputs, weights)])
    return summed_inputs

def cell(inputs, weights, threshold=0):
    summed_inputs = weighted_sum(inputs, weights)
    if summed_inputs > threshold:
        return 1
    else:
        return 0
```

```python
print(cell([1, 2, 3], [3, -2, 1], threshold=10))
print(cell([1, 2, 3], [3, -2, 1], threshold=0))
```
